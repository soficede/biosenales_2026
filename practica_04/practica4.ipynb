## 5.1 Etapas del registro de biopotenciales

Los biopotenciales son señales eléctricas generadas por la actividad electroquímica de células excitables, como las células nerviosas y musculares. Estas señales pueden propagarse a través de los tejidos y ser detectadas desde la superficie corporal mediante electrodos.

El proceso general para registrar una señal biomédica como ECG o EMG puede representarse mediante las siguientes etapas:

**Generación del potencial de acción → Propagación a través de los tejidos → Electrodos → Acondicionamiento de la señal → Conversión analógico-digital → Microcontrolador → Comunicación serial → Procesamiento y almacenamiento en el computador**

### Descripción de las etapas del registro de biopotenciales

1. **Generación del potencial de acción**  
   La señal eléctrica se origina en células excitables, como las células musculares y nerviosas. En estas células, los cambios en la distribución de iones a través de la membrana generan potenciales eléctricos. La suma de muchos de estos potenciales da origen a señales como el ECG y el EMG.

2. **Propagación a través de los tejidos**  
   Una vez generada la actividad eléctrica, esta se propaga a través de los tejidos del cuerpo. La señal que llega a la superficie corporal tiene una amplitud pequeña y puede verse afectada por la conductividad de los tejidos y por diferentes fuentes de interferencia.

3. **Electrodos**  
   Los electrodos permiten captar desde la superficie de la piel las diferencias de potencial generadas por la actividad biológica. Su correcta ubicación y el buen contacto con la piel son fundamentales para obtener una señal de buena calidad.

4. **Acondicionamiento de la señal**  
   La señal captada por los electrodos debe ser amplificada y acondicionada, ya que su amplitud es pequeña y puede contener ruido. En esta etapa se pueden realizar procesos de amplificación y filtrado para mejorar la señal antes de su digitalización.

5. **Conversión analógico-digital**  
   La señal proveniente del sistema de adquisición es originalmente analógica. Para poder procesarla en un computador, debe convertirse a valores digitales mediante un convertidor analógico-digital (ADC).

6. **Microcontrolador**  
   En esta práctica, el Arduino UNO recibe la señal analógica procedente del módulo de adquisición y realiza la conversión a valores digitales mediante su ADC.

7. **Comunicación serial**  
   Los valores adquiridos por el Arduino son enviados al computador a través del puerto serial. Esto permite transferir las muestras de la señal para su posterior procesamiento.

8. **Procesamiento y almacenamiento**  
   Finalmente, los datos recibidos en el computador pueden ser visualizados, analizados y almacenados. En esta práctica, Python fue utilizado para leer los datos enviados por el Arduino y guardarlos en archivos CSV para su posterior análisis.

### Componentes principales de un equipo de adquisición de biopotenciales

Un sistema de adquisición de biopotenciales está compuesto, de manera general, por los siguientes elementos:

- **Electrodos:** permiten captar las diferencias de potencial generadas por la actividad eléctrica del cuerpo. En señales de superficie como ECG y EMG, los electrodos se colocan sobre la piel.

- **Etapa de acondicionamiento analógico:** recibe la señal proveniente de los electrodos y permite adecuarla antes de su digitalización. Esta etapa puede incluir amplificación y filtrado para mejorar la relación señal-ruido.

- **Convertidor analógico-digital (ADC):** transforma la señal analógica acondicionada en valores digitales que pueden ser procesados por un sistema computacional.

- **Microcontrolador o sistema de adquisición:** se encarga de leer los valores digitales obtenidos a través del ADC y transmitirlos para su posterior análisis.

- **Sistema de comunicación:** permite enviar la información adquirida desde el sistema de adquisición hacia un computador. En esta práctica se utilizó comunicación serial.

- **Computador y software de procesamiento:** permiten visualizar, almacenar y procesar las señales adquiridas. En esta práctica se empleó Python para recibir los datos y guardarlos en archivos CSV.

En el desarrollo de la práctica, el código de adquisición en Python fue modificado para almacenar las señales adquiridas en formato CSV, lo que permitió conservar los datos y utilizarlos posteriormente en las etapas de procesamiento y análisis.

## 5.2 Fuentes de ruido en el registro de biopotenciales

Durante la adquisición de señales como ECG y EMG pueden aparecer diferentes fuentes de ruido o interferencia que alteran la señal registrada. Algunas permanecen relativamente constantes, mientras que otras cambian con el tiempo.

| Fuente de ruido | Tipo | Efecto sobre la señal | Estrategias de control |
|---|---|---|---|
| Interferencia de la red eléctrica (50/60 Hz) | Variante en el tiempo | Introduce una señal periódica superpuesta al ECG o EMG | Buena conexión a tierra, cables cortos, blindaje y filtro notch |
| Movimiento de los electrodos | Variante en el tiempo | Produce cambios bruscos y desplazamientos de la línea base | Fijar correctamente los electrodos, preparar la piel y evitar movimientos durante la adquisición |
| Mala impedancia piel-electrodo | Puede comportarse como relativamente estática o variar | Disminuye la calidad de la señal y aumenta la susceptibilidad al ruido | Limpiar la piel, retirar sudor o grasa y garantizar buen contacto del electrodo |
| Actividad muscular no deseada | Variante en el tiempo | Puede contaminar especialmente los registros ECG con señales EMG | Mantener al sujeto relajado y limitar movimientos durante la adquisición |
| Interferencia electromagnética de otros equipos | Variante en el tiempo | Puede inducir señales externas en los cables y electrodos | Alejar fuentes de interferencia, usar cables blindados y mejorar el aislamiento |
| Offset de los electrodos | Relativamente estática | Produce un nivel DC o desplazamiento de la línea base | Uso de amplificadores diferenciales y etapas de filtrado |
| Ruido electrónico del sistema de adquisición | Puede ser estático o aleatorio | Añade pequeñas fluctuaciones a la señal medida | Utilizar componentes adecuados, buena alimentación, filtrado y correcto diseño del sistema |

En términos generales, las fuentes de ruido estáticas o relativamente constantes generan desplazamientos o niveles de offset en la señal, mientras que las variantes en el tiempo producen interferencias que cambian durante el registro.

En bioseñales, las interferencias debidas al movimiento del paciente, la red eléctrica, la actividad muscular no deseada y otros equipos electrónicos son especialmente importantes, porque pueden tener amplitudes comparables o incluso superiores a la señal de interés.

### Relación de las fuentes de ruido con las señales adquiridas

En las señales registradas durante la práctica pueden presentarse alteraciones asociadas a diferentes fuentes de ruido. En el ECG, por ejemplo, pueden aparecer variaciones de la línea base debidas al movimiento, cambios en el contacto de los electrodos o respiración. También puede presentarse interferencia de la red eléctrica y contaminación por actividad muscular.

En la señal EMG, parte de la variabilidad observada corresponde a la propia activación muscular, pero también pueden aparecer artefactos debidos al movimiento de los electrodos, mala adherencia sobre la piel o interferencias externas.

Por esta razón, antes de analizar una bioseñal es importante evaluar su calidad y diferenciar qué componentes corresponden realmente a la actividad fisiológica y cuáles pueden deberse a ruido o artefactos de adquisición.

## 5.3 Acoplamiento piel-electrodo

El acoplamiento piel-electrodo corresponde a la interfaz mediante la cual el electrodo capta los potenciales eléctricos presentes en la superficie corporal. La calidad de este contacto influye directamente sobre la calidad de la señal registrada.

### Problemática asociada al acoplamiento piel-electrodo

La piel presenta una impedancia que puede dificultar la correcta captación de señales bioeléctricas, especialmente porque señales como ECG y EMG tienen amplitudes pequeñas. Además, factores como grasa, sudor, suciedad, vello o un mal contacto del electrodo pueden aumentar la impedancia de la interfaz y favorecer la aparición de ruido y artefactos.

También pueden aparecer diferencias de impedancia entre electrodos. Esto afecta la medición diferencial y hace que el sistema sea más sensible a interferencias externas.

### Estrategias para mejorar el acoplamiento

Para mejorar el contacto piel-electrodo se pueden aplicar varias estrategias:

- Limpiar previamente la piel para eliminar grasa, sudor y suciedad.
- Utilizar electrodos adecuados para registro superficial.
- Garantizar una buena adherencia del electrodo sobre la piel.
- Evitar que los cables ejerzan tensión sobre los electrodos.
- Reducir movimientos durante la adquisición.
- Colocar los electrodos en la ubicación anatómica adecuada.
- Verificar que todos los electrodos tengan un contacto similar con la piel.

### ¿Por qué es una etapa crítica?

El acoplamiento piel-electrodo es una de las etapas más críticas del registro de señales de superficie porque constituye el primer contacto entre la señal fisiológica y el sistema de adquisición.

Si la señal se capta de forma deficiente en esta etapa, el ruido, los artefactos o las pérdidas introducidas pueden propagarse a las etapas posteriores de amplificación, digitalización y procesamiento. Aunque después se apliquen filtros o algoritmos de procesamiento, no siempre es posible recuperar correctamente una señal que fue adquirida con mala calidad desde el inicio.

Por esta razón, un buen contacto piel-electrodo es fundamental para obtener registros confiables de ECG, EMG y otras bioseñales de superficie.

## 5.4 Análisis de la variabilidad de la frecuencia cardíaca (HRV)

Para realizar el análisis de HRV, primero se carga y visualiza la señal ECG adquirida durante la práctica. Antes de detectar los picos R, es necesario verificar que el archivo se lea correctamente y observar la forma general de la señal.

### Carga y visualización de ECG

```python
import pandas as pd
import matplotlib.pyplot as plt

# Cargar la señal ECG
ecg = pd.read_csv("ECG_torax.csv", header=None)

# Mostrar las primeras muestras
print(ecg.head())

# Mostrar información básica
print("Número de muestras:", len(ecg))

# Graficar la señal
plt.figure(figsize=(12, 4))
plt.plot(ecg[0])
plt.title("Señal ECG adquirida")
plt.xlabel("Muestra")
plt.ylabel("Valor ADC")
plt.grid()
plt.show()
```

### Frecuencia de muestreo

Para el procesamiento de la señal ECG es necesario conocer la frecuencia de muestreo utilizada durante la adquisición.

En el código de Arduino empleado en la práctica se utilizó un `delay(10)`, lo que introduce un intervalo aproximado de **10 ms** entre muestras.

Por lo tanto, el periodo de muestreo puede aproximarse como:

\[
T_s = 10\ \text{ms} = 0.01\ \text{s}
\]

y la frecuencia de muestreo como:

\[
f_s = \frac{1}{T_s} = \frac{1}{0.01} \approx 100\ \text{Hz}
\]

Por esta razón, para el análisis de la señal se asumió una frecuencia de muestreo aproximada de **100 Hz**.

Es importante tener en cuenta que este valor es una aproximación, ya que además del `delay(10)` existen pequeños tiempos adicionales asociados a la lectura analógica y a la transmisión serial de los datos.

```python
import numpy as np

fs = 100  # frecuencia de muestreo aproximada en Hz

# Convertir la columna a un arreglo de una dimensión
ecg_signal = ecg[0].to_numpy()

# Crear vector de tiempo
tiempo = np.arange(len(ecg_signal)) / fs

# Graficar ECG respecto al tiempo
plt.figure(figsize=(12, 4))
plt.plot(tiempo, ecg_signal)
plt.title("Señal ECG adquirida")
plt.xlabel("Tiempo (s)")
plt.ylabel("Valor ADC")
plt.grid()
plt.show()

print("Duración aproximada del registro:", len(ecg_signal) / fs, "segundos")
```

### Preprocesamiento de la señal

Al visualizar el ECG adquirido se observó que los complejos QRS presentaban una deflexión predominante negativa. Para facilitar la detección automática de los picos R, se invirtió la polaridad de la señal.

La inversión modifica únicamente la orientación vertical de la señal y no altera la ubicación temporal de los complejos QRS ni, por lo tanto, los intervalos RR.

```python
# Invertir la señal para facilitar la detección de los complejos QRS
ecg_invertido = -ecg_signal

plt.figure(figsize=(12, 4))
plt.plot(tiempo, ecg_invertido)

plt.title("Señal ECG invertida para detección de QRS")
plt.xlabel("Tiempo (s)")
plt.ylabel("Valor ADC invertido")
plt.grid()
plt.show()
```

```python
# Detección de picos R con NeuroKit2 sobre la señal invertida

signals_inv, info_inv = nk.ecg_process(ecg_invertido, sampling_rate=fs)

rpeaks_nk_inv = info_inv["ECG_R_Peaks"]

print("Número de picos R detectados con NeuroKit2:", len(rpeaks_nk_inv))
print("Posiciones de los picos R:", rpeaks_nk_inv)
```

```python
plt.figure(figsize=(12, 4))

plt.plot(tiempo, ecg_invertido, label="ECG invertido")

plt.scatter(
    rpeaks_nk_inv / fs,
    ecg_invertido[rpeaks_nk_inv],
    s=70,
    c="red",
    marker="o",
    label="Picos R detectados",
    zorder=3
)

plt.title("Detección de picos R con NeuroKit2 sobre ECG invertido")
plt.xlabel("Tiempo (s)")
plt.ylabel("Valor ADC invertido")
plt.legend()
plt.grid()
plt.show()
```

```python
# Detección con Pan-Tompkins sobre la señal invertida

detectors_inv = Detectors(fs)

rpeaks_det_inv = detectors_inv.pan_tompkins_detector(ecg_invertido)
rpeaks_det_inv = np.array(rpeaks_det_inv)

print("Número de picos R detectados con Pan-Tompkins:", len(rpeaks_det_inv))
print("Posiciones de los picos R:", rpeaks_det_inv)
```

```python
plt.figure(figsize=(12, 4))

plt.plot(tiempo, ecg_invertido, label="ECG invertido")

plt.scatter(
    rpeaks_det_inv / fs,
    ecg_invertido[rpeaks_det_inv],
    s=70,
    c="green",
    marker="x",
    label="Picos R - Pan-Tompkins",
    zorder=3
)

plt.title("Detección de picos R con Pan-Tompkins sobre ECG invertido")
plt.xlabel("Tiempo (s)")
plt.ylabel("Valor ADC invertido")
plt.legend()
plt.grid()
plt.show()
```

### Comparación de los algoritmos de detección de picos R 

Se compararon dos métodos para la detección de los picos R: NeuroKit2 y el algoritmo de Pan-Tompkins implementado mediante la librería `py-ecg-detectors`.

Debido a que los complejos QRS de la señal adquirida presentaban una deflexión predominantemente negativa, la señal fue invertida antes de realizar la comparación. Esta transformación no modifica la ubicación temporal de los latidos.

NeuroKit2 detectó **12 picos R**, ubicándolos de manera consistente sobre los principales complejos QRS observados en la señal.

Por otro lado, Pan-Tompkins detectó **13 eventos**. Aunque varios de ellos estuvieron asociados con complejos QRS reales, algunos marcadores presentaron un desplazamiento respecto al máximo del complejo y se observó una detección adicional al inicio del registro.

Por lo tanto, para la señal ECG adquirida en esta práctica, **NeuroKit2 presentó un mejor desempeño**, debido a que sus detecciones mostraron una mayor correspondencia visual con los complejos QRS y una menor presencia de detecciones adicionales.

```python
plt.figure(figsize=(12, 5))

plt.plot(tiempo, ecg_invertido, label="ECG invertido")

plt.scatter(
    rpeaks_nk_inv / fs,
    ecg_invertido[rpeaks_nk_inv],
    s=70,
    c="red",
    marker="o",
    label="NeuroKit2",
    zorder=3
)

plt.scatter(
    rpeaks_det_inv / fs,
    ecg_invertido[rpeaks_det_inv],
    s=70,
    c="green",
    marker="x",
    label="Pan-Tompkins",
    zorder=3
)

plt.title("Comparación de los algoritmos de detección de picos R")
plt.xlabel("Tiempo (s)")
plt.ylabel("Valor ADC invertido")
plt.legend()
plt.grid()
plt.show()
```

### Cálculo de HRV

A partir de los picos R detectados con NeuroKit2, se calculan los intervalos RR consecutivos en milisegundos. Con estos intervalos se obtiene una estimación de la frecuencia cardíaca y de la variabilidad entre latidos. Para continuar, se utiliza NeuroKit2 como método seleccionado en la comparación anterior.

```python
# Intervalos RR y frecuencia cardíaca
rpeaks = np.asarray(rpeaks_nk_inv)
rr_ms = np.diff(rpeaks) / fs * 1000

print("Intervalos RR (ms):", np.round(rr_ms, 2))
print("RR promedio (ms):", round(np.mean(rr_ms), 2))
print("Frecuencia cardíaca promedio (lpm):", round(60000 / np.mean(rr_ms), 2))
print("Frecuencia cardíaca por intervalo (lpm):", np.round(60000 / rr_ms, 2))
```

### Características de HRV en el dominio del tiempo

La función `get_time_domain_features` de `hrvanalysis` recibe los intervalos NN en milisegundos y calcula características temporales de la variabilidad cardíaca. En este registro se usan los intervalos RR detectados como aproximación de los intervalos NN; por ello, los resultados deben interpretarse con cautela, especialmente porque el registro contiene pocos latidos.

```python
from hrvanalysis import get_time_domain_features
from astropy.timeseries import LombScargle

hrv_tiempo = get_time_domain_features(nn_intervals=rr_ms.tolist())
pd.Series(hrv_tiempo, name="Valor").to_frame()

# La función recibe una lista de intervalos NN expresados en milisegundos
hrv_tiempo = get_time_domain_features(nn_intervals=rr_ms.tolist())

pd.Series(hrv_tiempo, name="Valor").to_frame()
```

**Interpretación:** el intervalo RR promedio permite estimar la frecuencia cardíaca media. SDNN representa la dispersión de los intervalos NN y RMSSD cuantifica las diferencias sucesivas entre intervalos; valores mayores indican más variabilidad en el registro analizado. Las métricas obtenidas son descriptivas y no permiten establecer conclusiones clínicas, ya que la señal es corta y no se ha realizado una validación manual ni una corrección específica de latidos ectópicos o detecciones erróneas.

### Aplicaciones del ECG

**1. Diagnóstico de isquemia e infarto agudo de miocardio.** El ECG de 12 derivaciones es la herramienta inicial para evaluar a un paciente con sospecha de síndrome coronario agudo. La presencia de elevación o depresión del segmento ST, ondas T anómalas o nuevas ondas Q permite identificar y localizar la isquemia, y es parte de los criterios para definir el infarto de miocardio [1]. El registro se hace con electrodos de superficie en las extremidades y el tórax, con el paciente en reposo, y su calidad depende de la preparación de la piel y de la ubicación correcta de los electrodos. En cuanto al procesamiento, las recomendaciones de estandarización establecen un ancho de banda diagnóstico (filtro paso alto cercano a 0,05 Hz para no distorsionar el segmento ST y un límite superior de aproximadamente 150 Hz), además de filtros para la interferencia de la red y la corrección de línea base [2].

**2. Detección de arritmias, en especial fibrilación auricular, con monitoreo ambulatorio.** Muchas arritmias son intermitentes y no aparecen en un ECG de reposo de pocos segundos. Por eso se usan dispositivos de registro prolongado, como el Holter o los parches adhesivos de ECG continuo. En un ensayo clínico aleatorizado, el monitoreo con un parche portable durante varias semanas permitió detectar más casos de fibrilación auricular que la atención habitual [3]. El registro es de una o pocas derivaciones, con electrodos adheridos al tórax durante días o semanas. El procesamiento se basa en detectar el complejo QRS, construir la serie de intervalos RR, identificar irregularidad del ritmo, rechazar segmentos con artefactos de movimiento y revisar los eventos marcados por el algoritmo. Esta aplicación está muy cerca de lo trabajado en la práctica, ya que depende de una buena detección de picos R.

**3. Evaluación de la función autonómica mediante la variabilidad de la frecuencia cardíaca (HRV).** La HRV describe las variaciones entre latidos consecutivos y refleja el balance entre la actividad simpática y parasimpática. Se ha utilizado para evaluar neuropatía autonómica y para estratificar el riesgo tras un infarto de miocardio [4]. El registro puede ser de corta duración (alrededor de 5 minutos, en reposo y en condiciones controladas) o de 24 horas con Holter. El procesamiento consiste en detectar los picos R, obtener la serie de intervalos RR, eliminar o corregir latidos ectópicos y artefactos, y calcular índices en el dominio del tiempo (SDNN, RMSSD) y de la frecuencia (bandas LF y HF). El documento de la Task Force recomienda frecuencias de muestreo de 250 a 500 Hz o mayores para localizar los picos R con precisión suficiente [4]. Esto es relevante para la práctica, porque la frecuencia de muestreo aproximada de 100 Hz usada aquí limita la resolución temporal de los intervalos RR.

### Aplicaciones del EMG

**1. Diagnóstico de enfermedades neuromusculares.** El EMG permite distinguir si una debilidad muscular tiene origen en el músculo (miopatía), en el nervio periférico (neuropatía), en la motoneurona o en la unión neuromuscular. Se utiliza en el estudio de radiculopatías, neuropatías y enfermedades de la motoneurona [5, 6]. El registro más habitual es el EMG de aguja, con un electrodo concéntrico introducido en el músculo, y se complementa con estudios de conducción nerviosa. El procesamiento incluye amplificación diferencial, filtrado en una banda amplia (en el orden de decenas de Hz hasta unos 10 kHz) y el análisis de la actividad de inserción, la actividad espontánea (como fibrilaciones), la forma, amplitud y duración de los potenciales de unidad motora y el patrón de reclutamiento durante la contracción [5, 6].

**2. Control de prótesis mioeléctricas.** En personas con amputación de miembro superior, el EMG de superficie registrado sobre los músculos remanentes se usa como señal de comando para mover una prótesis. Los sistemas más avanzados reconocen patrones de activación de varios canales para controlar más de un movimiento [7]. El registro es de superficie, con uno o varios canales dentro del encaje de la prótesis. El procesamiento incluye filtrado pasa banda (típicamente entre 20 y 450 Hz), rectificación y cálculo de la envolvente, extracción de características en ventanas de tiempo (por ejemplo, valor absoluto medio, longitud de onda o cruces por cero) y clasificación del movimiento deseado o estimación proporcional de la fuerza [7]. La posición de los electrodos y el contacto con la piel influyen mucho en la estabilidad del control [8].

**3. Rehabilitación y biofeedback electromiográfico.** En fisioterapia y rehabilitación, el EMG de superficie se usa para mostrar al paciente la activación de un músculo en tiempo real, por ejemplo, tras un accidente cerebrovascular, de modo que pueda aprender a activarlo o a relajarlo. Una revisión Cochrane evaluó el biofeedback con EMG para la recuperación motora después de un ACV [9]. El registro se hace con electrodos de superficie sobre el músculo entrenado, siguiendo recomendaciones de ubicación como las del proyecto SENIAM [8]. El procesamiento incluye filtrado, rectificación, cálculo de la envolvente o del valor RMS, comparación con un umbral establecido por el terapeuta y presentación del resultado como una señal visual o auditiva [10].

## Conclusiones
1. El registro de biopotenciales como el ECG y el EMG es una cadena de etapas, que va desde el potencial de acción y su propagación por los tejidos hasta los electrodos, el acondicionamiento, la conversión analógico-digital, la comunicación serial y el procesamiento en el computador. La calidad final depende de todas las etapas, y un fallo en alguna de ellas (por ejemplo, un mal contacto del electrodo) no se puede corregir por completo con el procesamiento posterior.

2. Las señales ECG y EMG tienen amplitudes pequeñas y son sensibles a ruido, como la interferencia de la red eléctrica, el movimiento de los electrodos, la impedancia piel-electrodo y la actividad muscular no deseada. Por esta razón, el acoplamiento piel-electrodo, la preparación de la piel y el control del movimiento son pasos críticos, y antes de interpretar una señal es necesario distinguir la actividad fisiológica de los artefactos.

3. En la señal ECG adquirida, los complejos QRS tenían polaridad predominantemente negativa, por lo que se invirtió la señal antes de detectar los picos R. Esta inversión no cambia la posición temporal de los latidos ni los intervalos RR. Al comparar los dos algoritmos, NeuroKit2 detectó 12 picos R con buena correspondencia visual con los complejos QRS, mientras que Pan-Tompkins detectó 13 eventos, algunos desplazados respecto al máximo del complejo y uno adicional al inicio del registro. Para esta señal, NeuroKit2 tuvo mejor desempeño, aunque este resultado se basa en una inspección visual de un único registro y no se puede generalizar.

4. La frecuencia de muestreo (aproximadamente 100 Hz, por el `delay(10)` del código de Arduino) es una limitación del sistema de adquisición usado. Es suficiente para observar el ritmo cardíaco, pero es baja para localizar con precisión los picos R en un análisis de HRV, donde se recomiendan frecuencias de 250 Hz o superiores [4]. Para un trabajo posterior convendría aumentar la frecuencia de muestreo y verificar que sea estable.

5. Las aplicaciones clínicas revisadas muestran que el tipo de registro y el procesamiento dependen del objetivo: el ECG de 12 derivaciones requiere un ancho de banda diagnóstico, el monitoreo ambulatorio se centra en la detección robusta de QRS, el EMG de aguja requiere un ancho de banda amplio y el EMG de superficie para prótesis o biofeedback se basa en la envolvente y en características de la señal [1-10].
