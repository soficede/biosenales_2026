```python
import numpy as np
import matplotlib.pyplot as plt
```

a) Creación de los vectores

```python
a = np.array([3.1, 1, -0.5, -3.2, 6])
b = np.array([1, 3, 2.2, 5.1, 1])

print("a =", a)
print("b =", b)
```

b) Multiplicación escalar (producto punto) de a·b

Para poder multiplicar dos vectores como producto escalar (dot product), **ambos vectores deben tener la misma longitud** (mismo número de elementos), ya que la operación consiste en multiplicar elemento a elemento y luego sumar los resultados. `a` y `b` ya cumplen esta condición (ambos tienen 5 elementos).

```python
producto_escalar = np.dot(a, b)
print("a · b =", producto_escalar)
```

c) Multiplicación punto a punto (elemento a elemento)

```python
producto_punto_a_punto = a * b
print("a .* b =", producto_punto_a_punto)
```

d) Construcción de la matriz A

```python
A = np.array([
    [2,    -1,   -3  ],
    [4,     1.5, -2.5],
    [7.3,  -0.9,  0.2]
])
print(A)
```

e) Transpuesta de A

```python
A_T = A.T
print(A_T)
```

f) Funciones de NumPy: ones, round, ceil, floor

```python
# ones: crea un arreglo lleno de unos
unos = np.ones((2, 3))
print("np.ones((2,3)) =\n", unos)

# round: redondea al entero más cercano (redondeo estándar)
redondeado = np.round(A)
print("np.round(A) =\n", redondeado)

# ceil: redondea hacia arriba (techo)
techo = np.ceil(A)
print("np.ceil(A) =\n", techo)

# floor: redondea hacia abajo (piso)
piso = np.floor(A)
print("np.floor(A) =\n", piso)
```

g) Elemento en la primera fila, tercera columna de A

```python
elemento = A[0, 2]
print("A[0,2] =", elemento)
```

h) Segunda fila de A

```python
segunda_fila = A[1, :]
print("Segunda fila de A =", segunda_fila)
```

i) Dimensiones de la matriz A

El comando para conocer las dimensiones de un arreglo en NumPy es `.shape`.

```python
print("Dimensiones de A:", A.shape)
```

j) Señal y[n] = sin(π·0.12·n), 0 ≤ n ≤ 100

```python
n = np.arange(0, 101)
y = np.sin(np.pi * 0.12 * n)
```

k) Señal y2[n] = cos(2π·0.03·n)

```python
y2 = np.cos(2 * np.pi * 0.03 * n)
```

l) Suma s[n] y producto t[n] de las dos señales

```python
s = y + y2
t = y * y2
```

m) Gráfica de y[n] y y2[n]

```python
plt.figure(figsize=(10, 4))
plt.plot(n, y, color='b', label='y[n] = sin(π·0.12n)')
plt.plot(n, y2, color='r', label='y2[n] = cos(2π·0.03n)')
plt.xlabel('n')
plt.ylabel('Amplitud')
plt.title('Señales y[n] y y2[n]')
plt.legend()
plt.grid(True)
plt.show()
```

n) Gráfica de s[n] y t[n]

```python
plt.figure(figsize=(10, 4))
plt.plot(n, s, color='g', label='s[n] = y[n] + y2[n]')
plt.plot(n, t, color='m', label='t[n] = y[n] · y2[n]')
plt.xlabel('n')
plt.ylabel('Amplitud')
plt.title('Señales s[n] y t[n]')
plt.legend()
plt.grid(True)
plt.show()
```

## Punto 4 — Impulso unitario

```python
def impseq(n0, n1, n2):
    # Genera x(n) = delta(n-n0); n1 <= n <= n2
    n = np.arange(n1, n2 + 1)
    x = (n - n0) == 0
    return [x, n]

x_imp, n_imp = impseq(0, -5, 5)
plt.stem(n_imp, x_imp)
plt.title('Impulso unitario δ(n)')
plt.xlabel('n')
plt.ylabel('δ(n)')
plt.grid(True)
plt.show()
```

p) Escalón unitario u(n − n0)

```python
def stepseq(n0, n1, n2):
    # Genera x(n) = u(n-n0); n1 <= n <= n2
    n = np.arange(n1, n2 + 1)
    x = (n - n0) >= 0
    return [x.astype(int), n]

x_u, n_u = stepseq(0, -5, 5)
plt.stem(n_u, x_u)
plt.title('Escalón unitario u(n)')
plt.xlabel('n')
plt.ylabel('u(n)')
plt.grid(True)
plt.show()
```

q) Rampa r(n − n0) con pendiente m

```python
def rampseq(n0, n1, n2, m=1):
    # Genera x(n) = m*(n-n0) para n >= n0, 0 en otro caso; n1 <= n <= n2
    n = np.arange(n1, n2 + 1)
    x = np.where(n >= n0, m * (n - n0), 0)
    return [x, n]

x_r, n_r = rampseq(0, -5, 5, m=1)
plt.stem(n_r, x_r)
plt.title('Rampa r(n), m = 1')
plt.xlabel('n')
plt.ylabel('r(n)')
plt.grid(True)
plt.show()
```
